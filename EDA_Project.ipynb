```python
histogram_columns = [
    'employee_count',
    'annual_revenue_million',
    'profit_margin_percent',
    'customer_count'
]

for col in histogram_columns:

    plt.figure(figsize=(8,5))

    sns.histplot(
        df_clean[col],
        bins=30,
        color='skyblue'
    )

    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')

    plt.show()

    print(f'Insight for {col}')
    print('The histogram helps understand the spread and distribution of data.')
    print('-'*80)
```

# 12. Outlier Detection Using IQR Method

### Outlier Detection and Analysis

In this section, outliers are identified using the Interquartile Range (IQR) method. Outliers are data points that lie significantly above or below the typical range of values in the dataset.

Outliers are important to analyze because they may represent exceptional business cases, such as very large organizations, companies with unusually high revenue, or businesses with exceptionally large customer bases.

According to the project requirements, outliers will be detected and analyzed but will not be removed, as they represent valid business observations rather than data entry errors.


```python
outlier_columns = [
    'employee_count',
    'annual_revenue_million',
    'profit_margin_percent',
    'customer_count'
]

for col in outlier_columns:

    Q1 = df_clean[col].quantile(0.25)

    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - (1.5 * IQR)

    upper_bound = Q3 + (1.5 * IQR)

    outliers = df_clean[
        (df_clean[col] < lower_bound) |
        (df_clean[col] > upper_bound)
    ]

    print(f"\nColumn: {col}")

    print(f"Number of Outliers: {outliers.shape[0]}")

    print(
        f"The IQR method detected {outliers.shape[0]} outliers in {col}."
    )

    print('-' * 80)
```

# 13. Bivariate Analysis

*   Numerical vs Numerical Analysis
*   Numerical vs Categorical Analysis



Bivariate Analysis examines the relationship between two variables.

The purpose is to understand whether changes in one variable influence another variable.

In this section, we will analyze:

• Numerical vs Numerical relationships
• Numerical vs Categorical relationships

These analyses help identify business patterns and important factors affecting company performance.

## Numerical vs Numerical Analysis

```python
scatter_pairs = [
    ('employee_count', 'annual_revenue_million'),
    ('customer_count', 'annual_revenue_million'),
    ('market_rating', 'profit_margin_percent')
]

for x_col, y_col in scatter_pairs:

    plt.figure(figsize=(8,5))

    sns.regplot(
        data=df_clean,
        x=x_col,
        y=y_col,
        scatter_kws={'alpha': 0.2},
        line_kws={'color': 'red'}
    )

    plt.title(f'{x_col} vs {y_col}', fontsize=14)

    plt.xlabel(x_col)

    plt.ylabel(y_col)

    plt.show()

    correlation = df_clean[x_col].corr(df_clean[y_col])

    print(f'\nInsight for {x_col} vs {y_col}')
    print(f'Correlation Value: {correlation:.2f}')

    if correlation >= 0.7:
        print('Strong positive relationship observed.')

    elif correlation >= 0.3:
        print('Moderate positive relationship observed.')

    elif correlation > -0.3:
        print('Weak or no significant relationship observed.')

    else:
        print('Negative relationship observed.')

    print('This analysis helps understand how changes in one variable are associated with changes in another variable.')

    print('-' * 80)
```

## Numerical vs Categorical Analysis


1.   Industry vs Revenue
2.   Region vs Employee Count
3.   Country vs Market Rating





### 1. Industry vs Revenue

```python
industry_revenue = df_clean.groupby(
    'industry'
)['annual_revenue_million'].mean().sort_values(
    ascending=False
)

plt.figure(figsize=(10,5))

industry_revenue.plot(
    kind='bar',
    color='skyblue'
)

plt.title('Average Revenue by Industry')

plt.xlabel('Industry')

plt.ylabel('Average Revenue')

plt.xticks(rotation=45)

plt.show()

print('Insight:')
print('This chart compares average revenue across industries.')
print('Industries with higher bars generate greater average revenue.')
```

### 2. Region vs Employee Count

```python
region_employee = df_clean.groupby(
    'region'
)['employee_count'].mean().sort_values(
    ascending=False
)

plt.figure(figsize=(10,5))

region_employee.plot(
    kind='bar',
    color='lightgreen'
)

plt.title('Average Employee Count by Region')

plt.xlabel('Region')

plt.ylabel('Average Employee Count')

plt.xticks(rotation=45)

plt.show()

print('Insight:')
print('This chart compares workforce size across regions.')
```

### 3. Country vs Market Rating

```python
country_rating = df_clean.groupby(
    'country'
)['market_rating'].mean().sort_values(
    ascending=False
)

plt.figure(figsize=(10,5))

country_rating.plot(
    kind='bar',
    color='salmon'
)

plt.title('Average Market Rating by Country')

plt.xlabel('Country')

plt.ylabel('Average Market Rating')

plt.xticks(rotation=45)

plt.show()

print('Insight:')
print('This chart compares average market ratings across countries.')
```

# 14. Correlation Matrix Analysis


*   Correlation Matrix
*   Heatmap






Correlation analysis helps measure the strength and direction of relationships between numerical variables.

The correlation coefficient ranges from -1 to +1.

• +1 indicates a perfect positive relationship.
• -1 indicates a perfect negative relationship.
• 0 indicates no relationship.

This analysis helps identify which business variables move together and which variables have little or no association.

```python
correlation_matrix = df_clean[numerical_columns].corr()

correlation_matrix
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')

plt.show()
```

### Interpretation

The heatmap visually displays the relationships between all numerical variables.

Values close to +1 indicate strong positive relationships.

Values close to -1 indicate strong negative relationships.

Values close to 0 indicate weak or no relationship.

This helps identify important business drivers and supports decision making.

# 15. Multivariate Analysis

Multivariate Analysis studies the relationship among three or more variables simultaneously.

This helps uncover deeper business patterns that cannot be observed through univariate or bivariate analysis alone.

In this section, we analyze how industry, region, and country influence revenue, employees, customers, profit margins, and market ratings.

### Analysis 1: Industry × Revenue × Employees

```python
industry_summary = df_clean.groupby('industry')[
    ['annual_revenue_million', 'employee_count']
].mean().sort_values(
    by='annual_revenue_million',
    ascending=False
)

industry_summary
```

```python
industry_summary.plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Industry vs Revenue and Employee Count')

plt.ylabel('Average Value')

plt.xticks(rotation=45)

plt.show()
```

### Insight

This analysis compares industries based on their average revenue and workforce size.

Industries with high revenue and high employee counts may represent large-scale operations, while industries with high revenue but fewer employees may indicate greater operational efficiency.

### Analysis 2: Region × Profit Margin × Market Rating

```python
region_summary = df_clean.groupby('region')[
    ['profit_margin_percent', 'market_rating']
].mean()

region_summary
```

```python
region_summary.plot(
    kind='bar',
    figsize=(10,6)
)

plt.title('Region vs Profit Margin and Market Rating')

plt.ylabel('Average Value')

plt.xticks(rotation=45)

plt.show()
```

### Insight

This analysis evaluates regional performance using profitability and customer perception.

Regions with higher profit margins and ratings may indicate stronger business environments and better customer satisfaction.

### Analysis 3: Country × Customers × Revenue

```python
country_summary = df_clean.groupby('country')[
    ['customer_count', 'annual_revenue_million']
].mean()

country_summary
```

```python
country_summary.plot(
    kind='bar',
    figsize=(12,6)
)

plt.title('Country vs Customers and Revenue')

plt.ylabel('Average Value')

plt.xticks(rotation=45)

plt.show()
```

### Insight

This analysis compares customer reach and revenue generation across countries.

Countries with larger customer bases and higher revenues may represent stronger market opportunities.
